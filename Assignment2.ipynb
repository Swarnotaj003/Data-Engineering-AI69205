**Task 2: Dictionaries**
1. Create a dictionary with student names as keys and their scores as values. The dictionary
should have at least 5 entries.
2. Calculate the average score and print it.
3. Find and print the student(s) with the highest score.

```python
students_dct = {}
students_dct['Swarnotaj'] = 92
students_dct['Ritesh'] = 82
students_dct['Arijit'] = 85
students_dct['Vinayak'] = 77
students_dct['Ngicheng'] = 80
print(students_dct)
```

```python
avg_score = 0

# accumulate sum from values
for score in students_dct.values():
    avg_score += score

# divide by the no. of students
avg_score /= len(students_dct)
print("Average score:", avg_score)
```

```python
# find the maximum key (name) based on its value (score) 
top_scorer = max(students_dct, key = lambda s: students_dct[s])
print("Highest scorer:", top_scorer)
```

**Task 6: DataFrame Manipulations**
1. Import the Pandas library.
2. Create a DataFrame with columns 'Name', 'Age', 'Department', 'Salary', and 'Years of
Service' with at least 7 entries.
3. Add a new column 'Bonus' which is 10% of the 'Salary' if 'Years of Service' is greater than 5,
otherwise 5%.

```python
import pandas as pd
```

```python
employees = {
    'Name': ['Arijit', 'Ritesh', 'Mani', 'Naveen', 'Vinayak', 'Swarnotaj', 'Ngicheng'],
    'Age': [28, 35, 42, 30, 26, 27, 38],
    'Department': ['HR', 'Engineering', 'Marketing', 'Finance', 'HR', 'Engineering', 'Marketing'],
    'Salary': [65000, 95000, 80000, 90000, 70000, 105000, 85000],
    'Years of Service': [3, 8, 12, 5, 2, 4, 7]
}
employees_df = pd.DataFrame(employees)
display(employees_df)
```

```python
# Fill the new column using np.where(condition, if value, else value)
employees_df['Bonus'] = np.where(employees_df['Years of Service'] > 5, 0.10, 0.05) * employees_df['Salary']
display(employees_df)
```

**Task 7: Handling Missing Data**
1. Introduce NaN values randomly into the DataFrame.
2. Fill NaN values in 'Salary' with the mean salary and drop rows where 'Department' is NaN.
3. Print the cleaned DataFrame.

```python
import random as rd

employees_df_with_nan = employees_df.copy()
rows, cols = employees_df_with_nan.shape

for i in range(7):
    # generate random address for the matrix (data frame)
    row = rd.randint(0, rows - 1)
    col = rd.randint(1, cols - 1) # exclude name
    employees_df_with_nan.iloc[row, col] = np.nan
    
display(employees_df_with_nan)
```

```python
# replace NaN salaries with avg salary
avg_salary = employees_df_with_nan['Salary'].mean()
employees_df_with_nan['Salary'] = round(employees_df_with_nan['Salary'].fillna(avg_salary), 2)

# drop rows with NaN Department in-place
employees_df_with_nan.dropna(subset=['Department'], inplace=True)
```

```python
display(employees_df_with_nan)
```

**Task 8: Grouping and Aggregation**
1. Group the DataFrame by 'Department' and calculate the average 'Salary' and total 'Years of
Service' for each department.
2. Print the grouped and aggregated data.

```python
aggregate_stats = employees_df.groupby('Department', as_index=False).agg(
    # aggregate value = ('column name', 'attribute')
    Average_Salary = ('Salary', 'mean'),
    Total_Years_of_Service = ('Years of Service', 'sum')
)
```

```python
display(aggregate_stats)
```
