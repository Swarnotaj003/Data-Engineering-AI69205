# Assignment 3

You are given a text file with two columns: 1st column is the key (string) and 2nd
column is the value (numerical). You need to write a python program from
scratch that will support the following functions (separate functions for each of
the operations):

**Task1**

Build a max heap based on the value field.

```python
class MaxHeap:
    def __init__(self, capacity):
        self.capacity = capacity
        self.idx = 0
        # arr stores [key, value] pair
        self.arr = [[] for _ in range(capacity)] 

    def size(self):
        return self.idx  

    def parent(self, i):
        return (i - 1) // 2

    def left(self, i):
        return 2 * i + 1

    def right(self, i):
        return 2 * i + 2

    def swap(self, i, j):
        self.arr[i], self.arr[j] = self.arr[j], self.arr[i]
 
    def maxHeapify(self, i):
        max = i
        left = self.left(i)
        right = self.right(i)
        # compare based on numeric value
        if left < self.idx and self.arr[left][1] > self.arr[max][1]:
            max = left
        if right < self.idx and self.arr[right][1] > self.arr[max][1]:
            max = right
        if max != i:
            self.swap(max, i)
            self.maxHeapify(max)   

    def insert(self, key, val):
        if self.idx >= self.capacity:
            raise OverflowError("Max heap is full!")
        self.arr[self.idx] = [key, val]
        self.idx += 1
        i = self.idx - 1
        # bubble up to maintain heap property
        while i > 0 and self.arr[self.parent(i)][1] < self.arr[i][1]:
            p = self.parent(i)
            self.swap(i, p)
            i = p

    def peekMax(self):
        if self.idx == 0:
            raise IndexError("Max heap is empty")
        maxVal = self.arr[0]
        return maxVal

    def extractMax(self):
        if self.idx == 0:
            raise IndexError("Max heap is empty")
        maxVal = self.arr[0]
        self.swap(0, self.idx - 1)
        self.idx -= 1
        self.maxHeapify(0)
        return maxVal
        
```

```python
# function to fetch data from text file into max heap
def fetchData(filename):
    maxheap = MaxHeap(int(1e6))
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            try:
                parts = line.split()
                if len(parts) < 2:
                    raise ValueError("Invalid line")
                key = parts[0]
                val = int(parts[1])
                # insert into the max heap
                maxheap.insert(key, val)
            except ValueError:
                print(f"Skipped the line with INVALID numeric value: {line}")
    return maxheap


# build the max heap
maxheap = fetchData('files/data.txt')
print("Size of the Max heap:", maxheap.size())
```

**Task 2**

Return the key associated with the k-th largest value.

```python
def getKthLargestKey(maxheap: MaxHeap, k: int):
    if k > maxheap.size():
        raise ValueError('k must not be larger than the size of heap')
    # extract max k times
    kmax = []
    for i in range(k):
        kmax = maxheap.extractMax()
    return kmax[0]


# print the kth largest key
try:
    k = int(input('Enter k: '))
    kthLargest = getKthLargestKey(maxheap, k)
    print(f"{k}th largest key: {kthLargest}")
except ValueError as ve:
    print(f"Value Error! {ve}")
```

**Task 3**

Sort the elements in descending order of values.

```python
def sortHeapDescending(maxheap: MaxHeap):
    sorted_data = []
    while maxheap.size() > 0:
        val = maxheap.extractMax()
        sorted_data.append(val)
    return sorted_data


# fetch max heap from the file
maxheap = fetchData('files/data.txt')

# sort the list in descending order  by value and show first few pairs
lst = sortHeapDescending(maxheap)
print("List sorted in descending order:")
print(lst[:20], "...")
```

**Task 4**

Return all the top k elements (based on value) under limited memory budget. 

In other words, you are not allowed to use a heap of size more than k elements.

```python
class MinHeap:
    def __init__(self, capacity):
        self.capacity = capacity
        self.idx = 0
        # arr stores [key, value] pair
        self.arr = [[] for _ in range(capacity)] 

    def size(self):
        return self.idx  

    def parent(self, i):
        return (i - 1) // 2

    def left(self, i):
        return 2 * i + 1

    def right(self, i):
        return 2 * i + 2

    def swap(self, i, j):
        self.arr[i], self.arr[j] = self.arr[j], self.arr[i]
 
    def minHeapify(self, i):
        min = i
        left = self.left(i)
        right = self.right(i)
        # compare based on numeric value
        if left < self.idx and self.arr[left][1] < self.arr[min][1]:
            min = left
        if right < self.idx and self.arr[right][1] < self.arr[min][1]:
            min = right
        if min != i:
            self.swap(min, i)
            self.minHeapify(min) 
        
    def insert(self, key, val):
        if self.idx >= self.capacity:
            raise OverflowError("Min heap is full!")
        self.arr[self.idx] = [key, val]
        self.idx += 1
        i = self.idx - 1
        # bubble up to maintain heap property
        while i > 0 and self.arr[self.parent(i)][1] > self.arr[i][1]:
            p = self.parent(i)
            self.swap(i, p)
            i = p
    
    def peekMin(self):
        if self.idx == 0:
            raise IndexError("Min heap is empty")
        minVal = self.arr[0]
        return minVal

    def extractMin(self):
        if self.idx == 0:
            raise IndexError("Min heap is empty")
        minVal = self.arr[0]
        self.swap(0, self.idx - 1)
        self.idx -= 1
        self.minHeapify(0)
        return minVal   
```

```python
# function to fetch only k largest data elements
def fetchKLargest(filename, k):
    minheap = MinHeap(k)
    with open(filename, 'r') as file:
        for line in file:
            line = line.strip()
            if not line:
                continue
            try:
                parts = line.split()
                if len(parts) < 2:
                    raise ValueError("Invalid line")
                key = parts[0]
                val = int(parts[1])
                # insert into the min heap without exceeding its size k
                if minheap.size() < k:
                    minheap.insert(key, val)
                else:
                    if val > minheap.peekMin()[1]:
                        minheap.extractMin()  
                        minheap.insert(key, val) 
            except ValueError:
                print(f"Skipped the line with INVALID numeric value: {line}")
    return minheap


# print all the k largest elements
try:
    k = int(input('Enter k: '))
    minheap = fetchKLargest('files/data.txt', k)
    print(f'{k} largest elements in ascending order')
    while minheap.size() > 0:
        print(minheap.extractMin())
except ValueError as ve:
    print(f"Value Error! {ve}")
```
